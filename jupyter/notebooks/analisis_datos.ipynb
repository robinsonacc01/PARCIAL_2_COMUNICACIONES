```python
import warnings
warnings.filterwarnings('ignore') # Oculta las advertencias informativas de Pandas

import psycopg2
import pandas as pd

# Datos de conexión a la base de datos PostgreSQL
DB_USER = "joomlauser"
DB_PASS = "joomlapassword123"
DB_HOST = "database"
DB_PORT = "5432"
DB_NAME = "joomladb"

try:
    # Conexión directa mediante psycopg2
    conn = psycopg2.connect(
        host=DB_HOST,
        port=DB_PORT,
        database=DB_NAME,
        user=DB_USER,
        password=DB_PASS
    )
    
    print("¡Conexión exitosa a la base de datos PostgreSQL desde Jupyter!\n")
    
    # Consulta directa a las tablas
    query = "SELECT table_name FROM information_schema.tables WHERE table_schema='public';"
    df_tables = pd.read_sql_query(query, conn)
    
    print(f"Se encontraron {len(df_tables)} tablas creadas por Joomla:")
    print(df_tables.head(10))
    
    # Cerrar conexión
    conn.close()

except Exception as e:
    print("Error al conectar con la base de datos:", e)
```
